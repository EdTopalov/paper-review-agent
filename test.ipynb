```python
from paper_review.llm import OpenAILLM
llm = OpenAILLM()

result = llm.ask([
    {
        "role": "system",
        "content": "Return JSON only."
    },
    {
        "role": "user",
        "content": "Return a JSON object with the answer 42."
    }
])

print(result["raw"])
```

```python
from paper_review.paper import load_paper

paper = load_paper("data/papers/LORA.pdf")

print("Title:", paper.title)
print("Pages:", len(paper.pages))
print("Opening type:", paper.opening_label)
print("Opening:", paper.opening[:2000])
```

```python
from paper_review.runner import run_baseline
import json

baseline_result = run_baseline(paper=paperб llm=llm, trace_dir="traces")

print(json.dumps(baseline_result, indent=2, ensure_ascii=False))
```

```python
from paper_review.runner import run_agent

agent_result = run_agent(
    paper=paper,
    llm=llm,
    trace_dir="traces",
    max_tool_calls=3,
    max_model_calls=4,
)

print(json.dumps(agent_result, indent=2, ensure_ascii=False))
```
