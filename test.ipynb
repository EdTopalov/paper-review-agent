# Paper review agent: optional notebook example
The CLI in M2.md is the documented entry point. These cells use the same OpenAI runner and frozen inputs. Run from the repository root after installing the package; keep the API key in the ignored .env file. No notebook output is an evaluation result.

```python
from pathlib import Path
from datetime import datetime
import json
import os
from dotenv import load_dotenv
from paper_review.paper import Paper
from paper_review.runner import Config, run_review

load_dotenv(".env", override=True)
config = Config(model=os.environ.get("OPENAI_MODEL", "gpt-6-luna"))
paper_path = Path("data/eval/papers/T01.json")
output = Path("runs/scratch") / ("notebook-" + datetime.now().strftime("%Y%m%d-%H%M%S"))
```

```python
paper = Paper.load(paper_path)
print(paper.snapshot["title"])
print("Pages:", len(paper.pages))
```

```python
baseline_result = run_review(paper_path, output / "baseline", config, "T01", variant="baseline")
print(json.dumps(baseline_result, indent=2))
```

```python
agent_result = run_review(paper_path, output / "agent", config, "T01", variant="agent")
print(json.dumps(agent_result, indent=2))
```
